```python
#dataset is from data frame
# collection of data in row column format

#dataframe
# row ,column format into python

# series are the building block of dataframe

# difference btw array and series

# array=> implict ,less memory, faster , n dimension (1d,2,3d.....)
# series=> explicit index,more memory,slower, 1 dimension
import numpy as np
```

```python
import pandas as pd
import numpy as np

a=np.arange(4,10)
s=pd.Series(a)
print(s)

print(s.ndim)
print(s.size)
print(s.shape)

print(s.head(3))
print(s.tail(3))
```

```python
# create  two series
a=pd.Series([4,5,6,7,8])
b=pd.Series([5,4,3,6,7])
print(a)
print(b)
d=pd.concat([a,b],ignore_index=True)   #concat is used to combine series
print(d)                               #add a parameter ignore index
```

```python
e=a.add(b)
print(e)

e=a.sub(b)
print(e)

e=a.mul(b)
print(e)

e=a.div(b)
print(e)
```

```python
z=pd.Series([3,4,5,6],index=["a","b","c","d"])  # for custom index
z
```

```python
d={"id":1,"age":54,"class":"ds"}
z=pd.Series(d)
z
```

```python
z=pd.Series(4,index=[1,2,3,4])   #to store same value in position
z
```

```python
z=pd.Series([1,2,3,4,5,6,7])
print(z)

z[3]
z[1:4]
z[[6,1,5]]  # to fetch multiple element use [[]](use double square brackets)
```

```python
#dataframe (2d)
#create dataframe from list

l=[1,2,3,4,5]

df=pd.DataFrame(l)

df.shape
df.size
df.ndim
print(df)
```

```python
#create dataframe from nested list
l=[["apple",102],["orange",48],["grapes",56]]
df=pd.DataFrame(l,columns=["fruits","price"])
df
```

```python
#create a dataframe with id name location and add 4 rows
l=[[11,"akash","TVM"],[12,"nabil","TVM"],[13,"farhan","Tvm"],[14,"karthik","TVM"]]
df=pd.DataFrame(l,columns=["Id","Name","location"])
df
print(df.head(1))
print(df.tail(2))
df.head(2)
```

```python
import pandas as pd

#create dataframe from list of dictionay

l=[{"fruits":"apple","price":234},{"fruits":"orange","price":87},{"fruits":"grapes","price":90}]
df=pd.DataFrame(l)
df
```

```python
# create data frame with id name age salary profession and add 5 values
d={"ID":pd.Series([1,2,3,4,5]),"Name":pd.Series(["Akash","Nabil","Farhan","Rihan","Adithyan"]),"Age":pd.Series([22,22,23,24,25]),"Salary":pd.Series([15000,16000,15000,14000,12000]),"Profession":pd.Series(["IT","Developer","Analyst","Java","Python"])}
df=pd.DataFrame(d)
df

df["location"]=["tvm","kochi","kollam","tvm","kochi"]
df
```
